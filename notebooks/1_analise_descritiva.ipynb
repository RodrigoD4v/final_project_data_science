## 9.2 Gráficos

```python
df["country"].value_counts().head(10).sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Top 10 países por quantidade de registros")
plt.xlabel("Quantidade de registros")
plt.ylabel("País")
plt.show()
```

```python
df["employmentType"].fillna("Não Informado").value_counts().plot(kind="bar")

plt.title("Distribuição por tipo de emprego")
plt.xlabel("Tipo de emprego")
plt.ylabel("Quantidade de registros")
plt.show()
```

```python
df["status"].fillna("Não Informado").value_counts().plot(kind="bar")

plt.title("Distribuição dos registros por status")
plt.xlabel("Status")
plt.ylabel("Quantidade de registros")
plt.show()
```

```python
df['onboarding_status'].fillna("Não Informado").value_counts().plot(kind='bar')

plt.title("Distribuição dos registros por status de onboarding")
plt.xlabel("Status de onboarding")
plt.ylabel("Quantidade de registros")
plt.show()
```

```python
dados_grafico = pd.crosstab(df['status'].reset_index(drop=True), df['payFrequency'].reset_index(drop=True), dropna=False)

dados_grafico.plot(kind='bar', stacked=True, figsize=(10, 6), cmap='viridis')

plt.title('Distribuição da Frequência de Pagamento por Status do Contrato')
plt.xlabel('Status do Contrato')
plt.ylabel('Quantidade de Registros')
plt.xticks(rotation=45)
plt.legend(title='Frequência de Pagamento')
plt.tight_layout()
plt.show()
```
